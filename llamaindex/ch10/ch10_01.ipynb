10.1 MCP
에이전트(Agent) 다양한 기능, 서비스, 시스템과 안전하고 일관되게 상호작용 할 수 있도록 설계된 도구 호출 표준 프로토콜
- 툴(Tool) : 에이전트가 호출할 수 있는 기능 단위, 이름, 설명, 입력 파라미터 반환 값 등을 JSON 형식으로 정의
            모델이 어떤 기능을 어떤 방식으로 사용할 수 있는지 이해하는데 필요한 정보
- MCP 클라이언트 : 모델 또는 애플리케이션 측에서 MCP 서버의 툴을 호출하는 인터페이스 역할을 수행
- MCP 서버 : 정의된 툴을 실제로 실행하고 결과를 반환하는 실행 주체로, 외부 API 또는 로컬 도구와 연결 됨
- 통신 규약(JSON-RPC 2.0) : 모델과 MCP 서버 간의 툴 호출 요청 및 응답을 주고받는 데 사용되는 메세지 포맷

- 툴이라는 표준화된 인터페이스를 중심으로 아키텍처를 구조화
  -> 다양한 도구와 애플리케이션 간의 상호작용이 표준화되고 재사용 가능해지며, 확장성 또한 획기적으로 향상됨

10.3.1 MCP 서버
- 다양한 도구, 시스템, 리소스를 MCP표준에 따라 래핑하는 역할
10.3.2 MCP 인스펙터(Inspector)
- 웹 기반 UI를 제공하는 개발 도구, MCP 서버를 테스트하고 디버깅 하기 위한 개발 도구
10.3.3 메세지 형식
- 요청/응답/알림 메세지
1. 표준 입출력 : 로컬환경에서 CLI기반으로 MCP 서버와 통신할 때 사용
   mcp.run(transport="studio")
2. 서버 전송 : HTTP 기반의 비동기 스트리밍 방식, 서버에서 클라이언트로 실시간 메시지를 푸쉬할 수 있는 기능 제공
   mcp.run(transport="sse")

10.4 MCP 클라이언트
- AI 애플리케이션 내부에 존재하면서, 사용자의 요청이나 모델의 내부판단에 따라 외부 MCP 서버에 명령을 전달하는 역할
